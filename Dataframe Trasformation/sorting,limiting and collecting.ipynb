```python
flight_time_df = spark.read.table("dev.spark_db.flight_time")
```

```python
flight_time_df.display()
```

```python
from pyspark.sql.functions import col,desc

flight_time_df.filter((col("FL_DATE")=='2000-01-16')
                      & (col("ORIGIN")=='AUS')
                      & (col("DEST")=='ORD')).withColumn('duration',col("ARR_TIME")-col("DEP_TIME")).orderBy(desc("duration")).limit(3).display()
```
