```python
person_list = [
    (100,"Prashant",30),
    (101,"David",None),
    (102,"sushant",None),
    (103,"Abdul",45),
    (104,"Shruti",28)
]

person_df = spark.createDataFrame(person_list).toDF("id","name","age")
display(person_df)
```

find record where age is 28
find all record where age is not given or unknown

```python
from pyspark.sql.functions import col,expr
age_df = person_df.filter(col("age").isNull())
age_df.display()
```

```python
person_df.where(col("age").isNull()).display()
person_df1 = person_df.withColumn("age_null_exprt",expr("age is not null")).display()
```

how operators work on nulls


```python
person_df.withColumn("age_ft_28",expr("age > 28")).display()
```

```python
person_df.where("age is not null").withColumn("exprience",expr("nvl(age,10) -10")).display()
```
