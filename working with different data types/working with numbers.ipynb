```python
from pyspark.sql.function import col,expr
```

```python
retail_df = (
    spark.read.format("csv")
    .option("header","true")
    .option("inferschema","true")
    .load("/Volumes/dev/spark_db/datasets/spark_programming/data/invoices.csv")
)
```

```python
retail_df.limit(6).display()
```

```python
from pyspark.sql.functions import expr, col

total_value_df = retail_df.withColumn("total_value",expr("quantity * unitprice"))
total_value_df.display()                                                      
```

```python
from pyspark.sql.functions import round
cal_logic = round(col("Quantity")* col("Quantity"),2)
total_value_df_1 = retail_df.withColumn("total_value",cal_logic)
total_value_df_1.display()
```
