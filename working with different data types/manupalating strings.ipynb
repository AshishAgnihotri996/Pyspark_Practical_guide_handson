create a date from day month and year


```python
df= spark.createDataFrame([("25","07","1982")]).toDF("day","month","year")
df.display()
```

```python
from pyspark.sql.functions import concat_ws,to_date
data_expr= to_date(concat_ws("-","year", "month", "day"))

df.select(data_expr.alias("date")).display()
```

create two fields as fun text and salary

```python
from pyspark.sql.functions import concat_ws,to_date,format_number,format_string

df = (spark.createDataFrame([("sanjay","karla",25,"july",1982,19408.98)])
      .toDF("first_name","last_name","day","month","year","salary"))

salary_fmt = format_number("salary","$###,###.##")

text_fmt = format_string("Mr. %s %s was born on %th %s of %d.","first_name","last_name","day","month","year","salary").alias("fun_text")

df.select(salary_fmt,text_fmt).display()
```
